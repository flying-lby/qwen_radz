```python
import json

# 原始文件路径
input_path = "/home/lby/iclr2026/qwen_radz/qwen_finetune/Qwen2-VL-Finetune/data/new_classify_mimic_file_clip.json"
# 新文件路径
output_path = "/home/lby/iclr2026/qwen_radz/qwen_finetune/Qwen2-VL-Finetune/train_data_100.json"

# 读取前100条数据
with open(input_path, 'r', encoding='utf-8') as f:
    data = json.load(f)

# 截取前100条
subset = data[:100]

# 保存为新文件
with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(subset, f, indent=2, ensure_ascii=False)

print(f"保存成功，共 {len(subset)} 条数据。路径: {output_path}")
```
